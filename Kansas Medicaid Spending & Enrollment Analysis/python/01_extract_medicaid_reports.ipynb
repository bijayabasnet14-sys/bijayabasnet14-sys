```python
# Open the FY2017 Medicaid PDF using pdfplumber
with pdfplumber.open(pdf_path) as pdf:
    
    # Access page 4 of the PDF.
    # Python uses zero-based indexing, so index 3 represents the fourth page.
    page = pdf.pages[3]
    
    # Extract all readable text from the page
    text = page.extract_text()

# Split the extracted text into individual lines
# "\n" represents a new line in the extracted PDF text
lines = text.split("\n")

# Loop through each extracted line
for line in lines:
    
    # repr() shows the exact text stored in each line,
    # which helps us inspect the structure before cleaning the data
    print(repr(line))
```

```python
# Import Python's regular expression module.
# We will use regex to recognize and extract values from each monthly line.
import re

# Define the 12 month names we expect to find in the report.
months = [
    "July", "August", "September", "October",
    "November", "December", "January", "February",
    "March", "April", "May", "June"
]

# Create an empty list where we will store each extracted monthly record.
monthly_records = []

# Loop through every line extracted from the PDF.
for line in lines:
    
    # Check whether the line begins with one of our month names.
    # Only monthly data rows should pass this test.
    if any(line.startswith(month) for month in months):
        
        # Use a regular expression to capture:
        # 1. Month name
        # 2. Number of weeks shown in parentheses
        # 3. Beneficiaries
        # 4. Beneficiary average cost
        # 5. Consumers
        # 6. Consumer average cost
        # 7. Total expenditures
        pattern = (
            r"^(\w+) "
            r"\((\d+) week\) "
            r"([\d,]+) "
            r"\$([\d,.]+) "
            r"([\d,]+) "
            r"\$([\d,.]+) "
            r"\$([\d,]+)"
        )

        match = re.search(pattern, line)

        # If the line matches the expected structure,
        # extract the values and save them as a dictionary.
        if match:
            record = {
                "fiscal_year": 2017,
                "month": match.group(1),
                "weeks": int(match.group(2)),
                "beneficiaries": int(match.group(3).replace(",", "")),
                "beneficiary_avg_cost": float(match.group(4).replace(",", "")),
                "consumers": int(match.group(5).replace(",", "")),
                "consumer_avg_cost": float(match.group(6).replace(",", "")),
                "total_expenditures": int(match.group(7).replace(",", ""))
            }

            monthly_records.append(record)

# Display the extracted records so we can verify the parser worked correctly.
monthly_records
```

```python
# Create a reusable function that extracts the monthly
# "Beneficiaries and Expenditures" data from one fiscal-year PDF.

def extract_medicaid_year(pdf_path, fiscal_year):
    
    # Open the PDF file.
    with pdfplumber.open(pdf_path) as pdf:
        
        # Start with no matching page found.
        target_page_text = None
        
        # Search every page for the section title.
        for page in pdf.pages:
            
            # Extract text from the current page.
            # Use an empty string if the page contains no readable text.
            page_text = page.extract_text() or ""
            
            # Identify the page containing the monthly
            # Beneficiaries and Expenditures table.
            if "Beneficiaries and Expenditures" in page_text:
                target_page_text = page_text
                break

    # Stop with an error if the expected section was not found.
    # This protects us from silently extracting incorrect data.
    if target_page_text is None:
        raise ValueError(
            f"Beneficiaries and Expenditures section not found for FY{fiscal_year}"
        )

    # Split the page into individual text lines.
    lines = target_page_text.split("\n")

    # Define the month names expected in each fiscal-year report.
    months = [
        "July", "August", "September", "October",
        "November", "December", "January", "February",
        "March", "April", "May", "June"
    ]

    # Create an empty list to store the extracted monthly records.
    monthly_records = []

    # Define the structure expected for each monthly row.
    # The pattern captures:
    # month, number of weeks, beneficiaries, beneficiary average cost,
    # consumers, consumer average cost, and total expenditures.
    pattern = (
        r"^(\w+) "
        r"\((\d+) week\) "
        r"([\d,]+) "
        r"\$([\d,.]+) "
        r"([\d,]+) "
        r"\$([\d,.]+) "
        r"\$([\d,]+)"
    )

    # Examine each line from the selected page.
    for line in lines:

        # Only process rows that begin with a month name.
        if any(line.startswith(month) for month in months):

            # Try to match the line against the expected monthly pattern.
            match = re.search(pattern, line)

            # If the pattern matches, convert the text values
            # into clean numeric fields.
            if match:
                record = {
                    "fiscal_year": fiscal_year,
                    "month": match.group(1),
                    "weeks": int(match.group(2)),
                    "beneficiaries": int(match.group(3).replace(",", "")),
                    "beneficiary_avg_cost": float(match.group(4).replace(",", "")),
                    "consumers": int(match.group(5).replace(",", "")),
                    "consumer_avg_cost": float(match.group(6).replace(",", "")),
                    "total_expenditures": int(match.group(7).replace(",", ""))
                }

                monthly_records.append(record)

    # Convert the extracted monthly records into a pandas DataFrame.
    df = pd.DataFrame(monthly_records)

    # Return the completed fiscal-year dataset.
    return df
```

```python
# Create a reusable validation function for one fiscal-year DataFrame.
# This checks the basic structure of the extracted data before we trust it.

def validate_medicaid_year(df, fiscal_year):
    
    # Expected fiscal-year month order.
    expected_months = [
        "July", "August", "September", "October",
        "November", "December", "January", "February",
        "March", "April", "May", "June"
    ]
    
    # Check 1: confirm that exactly 12 monthly rows were extracted.
    row_check = len(df) == 12
    
    # Check 2: confirm that the months appear in the expected order.
    month_check = df["month"].tolist() == expected_months
    
    # Check 3: confirm that no values are missing.
    missing_check = df.isna().sum().sum() == 0
    
    # Display the results clearly.
    print(f"FY{fiscal_year} validation")
    print("-" * 25)
    print("12 rows:", row_check)
    print("Correct months:", month_check)
    print("No missing values:", missing_check)
    
    # Calculate annual totals that we can later compare
    # against the official cumulative totals in the report.
    annual_expenditures = df["total_expenditures"].sum()
    average_beneficiaries = df["beneficiaries"].mean()
    
    print(f"\nAnnual expenditures: ${annual_expenditures:,.0f}")
    print(f"Average monthly beneficiaries: {average_beneficiaries:,.0f}")
```

```python
# Create an empty list to store each fiscal year's DataFrame.
all_years = []

# Loop through fiscal years 2017 through 2026.
for fiscal_year in range(2017, 2027):
    
    # Build the file path for the current fiscal-year PDF.
    pdf_path = Path(f"../data/raw/MAR FY {fiscal_year}.pdf")
    
    # Extract the monthly Medicaid data using our reusable function.
    yearly_df = extract_medicaid_year(pdf_path, fiscal_year)
    
    # Store the extracted DataFrame in our list.
    all_years.append(yearly_df)

# Combine all 10 fiscal-year DataFrames into one dataset.
# ignore_index=True creates a clean row index from 0 onward.
medicaid_monthly_df = pd.concat(all_years, ignore_index=True)

# Display the combined dataset.
medicaid_monthly_df
```

```python
# ---------------------------------------------------------
# CREATE ANNUAL MEDICAID SUMMARY
# ---------------------------------------------------------

# Aggregate the 120 monthly records into one row per fiscal year.
#
# We use:
# - mean() for beneficiaries to calculate average monthly enrollment
# - mean() for consumers to calculate average monthly consumers
# - sum() for beneficiaries to calculate total beneficiary-months
# - sum() for expenditures to calculate total annual spending
annual_summary = (
    medicaid_monthly_df
    .groupby("fiscal_year")
    .agg(
        avg_monthly_beneficiaries=("beneficiaries", "mean"),
        avg_monthly_consumers=("consumers", "mean"),
        beneficiary_months=("beneficiaries", "sum"),
        total_expenditures=("total_expenditures", "sum")
    )
    .reset_index()
)


# Calculate expenditure per beneficiary-month.
#
# A beneficiary-month represents one beneficiary enrolled for one month.
# Dividing annual expenditures by total beneficiary-months gives us
# a normalized measure of spending relative to enrollment.
#
# This is a spending-intensity measure. It should not be interpreted
# as an individual's medical cost or as a measure of program efficiency.
annual_summary["expenditure_per_beneficiary_month"] = (
    annual_summary["total_expenditures"]
    / annual_summary["beneficiary_months"]
)


# Calculate year-over-year percentage change in average monthly beneficiaries.
# pct_change() compares each fiscal year with the previous fiscal year.
annual_summary["beneficiary_yoy_pct"] = (
    annual_summary["avg_monthly_beneficiaries"]
    .pct_change()
    * 100
)


# Calculate year-over-year percentage change in total expenditures.
annual_summary["expenditure_yoy_pct"] = (
    annual_summary["total_expenditures"]
    .pct_change()
    * 100
)


# Round average monthly counts to whole beneficiaries/consumers.
annual_summary["avg_monthly_beneficiaries"] = (
    annual_summary["avg_monthly_beneficiaries"].round(0)
)

annual_summary["avg_monthly_consumers"] = (
    annual_summary["avg_monthly_consumers"].round(0)
)


# Round expenditure per beneficiary-month to cents.
annual_summary["expenditure_per_beneficiary_month"] = (
    annual_summary["expenditure_per_beneficiary_month"].round(2)
)


# Round year-over-year percentage changes to two decimal places.
annual_summary["beneficiary_yoy_pct"] = (
    annual_summary["beneficiary_yoy_pct"].round(2)
)

annual_summary["expenditure_yoy_pct"] = (
    annual_summary["expenditure_yoy_pct"].round(2)
)


# Display the completed annual summary for validation.
annual_summary
```

```python
# ---------------------------------------------------------
# FINAL QUALITY CHECKS FOR THE 10-YEAR MONTHLY DATASET
# ---------------------------------------------------------

# Check 1: Look for duplicate fiscal year + month combinations.
# Each fiscal year should contain exactly one record for each month.
duplicate_rows = medicaid_monthly_df.duplicated(
    subset=["fiscal_year", "month"]
).sum()

print("Duplicate fiscal year/month rows:", duplicate_rows)


# Check 2: Look for missing values anywhere in the dataset.
missing_values = medicaid_monthly_df.isna().sum().sum()

print("Total missing values:", missing_values)


# Check 3: Check for zero or negative beneficiary counts.
# These would be suspicious for monthly Medicaid enrollment data.
invalid_beneficiaries = (
    medicaid_monthly_df["beneficiaries"] <= 0
).sum()

print("Invalid beneficiary values:", invalid_beneficiaries)


# Check 4: Check for zero or negative total expenditures.
# Unexpected values here would require us to return to the source PDF.
invalid_expenditures = (
    medicaid_monthly_df["total_expenditures"] <= 0
).sum()

print("Invalid expenditure values:", invalid_expenditures)


# Check 5: Determine whether consumers ever exceed beneficiaries.
# Beneficiaries represent eligible individuals, while consumers
# represent individuals associated with paid services.
consumer_over_beneficiary = (
    medicaid_monthly_df["consumers"]
    > medicaid_monthly_df["beneficiaries"]
).sum()

print(
    "Months where consumers exceed beneficiaries:",
    consumer_over_beneficiary
)


# Check 6: Confirm the complete fiscal-year range.
print(
    "Fiscal year range:",
    medicaid_monthly_df["fiscal_year"].min(),
    "to",
    medicaid_monthly_df["fiscal_year"].max()
)


# Check 7: Confirm the final number of observations.
print("Total monthly observations:", len(medicaid_monthly_df))
```

```python
# ---------------------------------------------------------
# FUNCTION: EXTRACT CATEGORY-OF-SERVICE ANNUAL TOTALS
# ---------------------------------------------------------

def extract_category_expenditures(pdf_path, fiscal_year):
    """
    Extract annual Category of Service expenditure totals
    from Kansas Medicaid reports using the FY2018-FY2026 format.
    """

    # Open the Medicaid report.
    with pdfplumber.open(pdf_path) as pdf:

        # Page 10 contains the Category of Service table
        # for FY2018-FY2026.
        page_text = pdf.pages[9].extract_text() or ""

    # Store the extracted category records here.
    records = []

    # Process the table one line at a time.
    for line in page_text.splitlines():

        # Extract all monetary values from the current row.
        # Negative values are allowed because some report
        # categories contain accounting adjustments.
        dollar_values = re.findall(
            r"-?\$[\d,]+(?:\.\d+)?",
            line
        )

        # A complete category row should contain:
        # 12 months + Total + Average + State Funds = 15 values.
        if len(dollar_values) == 15:

            # Find where the first monetary value begins.
            first_dollar = re.search(r"-?\$", line)

            # Everything before the first dollar value
            # represents the category name.
            category = line[:first_dollar.start()].strip()

            # Index 12 is the annual Total column.
            annual_total_text = dollar_values[12]

            # Convert the formatted dollar string into an integer.
            # Example: "$1,234,567" becomes 1234567.
            annual_total = int(
                annual_total_text
                .replace("$", "")
                .replace(",", "")
            )

            # Add the clean record to our results.
            records.append({
                "fiscal_year": fiscal_year,
                "category": category,
                "annual_expenditures": annual_total
            })

    # Return the extracted records as a DataFrame.
    return pd.DataFrame(records)
```

```python
# ---------------------------------------------------------
# EXTRACT CATEGORY DATA FOR FY2018-FY2026
# ---------------------------------------------------------

# Store each fiscal year's extracted DataFrame.
all_category_years = []

for fiscal_year in range(2018, 2027):

    # Construct the appropriate PDF filename automatically.
    pdf_path = Path(
        f"../data/raw/MAR FY {fiscal_year}.pdf"
    )

    # Extract this year's category expenditures.
    yearly_categories = extract_category_expenditures(
        pdf_path,
        fiscal_year
    )

    # Store it for later combination.
    all_category_years.append(yearly_categories)

    # Print the number of rows extracted as a quick
    # structural check for each fiscal year.
    print(
        f"FY{fiscal_year}: "
        f"{len(yearly_categories)} rows extracted"
    )

# Combine all nine fiscal years into one dataset.
category_df = pd.concat(
    all_category_years,
    ignore_index=True
)

print("\nCombined shape:", category_df.shape)
```

```python
# ---------------------------------------------------------
# VALIDATE CATEGORY-OF-SERVICE TOTALS
# ---------------------------------------------------------

# Select only the "Total Expenditures" row from each fiscal year.
# This gives us the annual total reported on the
# Category of Service page.
category_totals = (
    category_df[
        category_df["category"] == "Total Expenditures"
    ][
        ["fiscal_year", "annual_expenditures"]
    ]
    .rename(
        columns={
            "annual_expenditures": "category_report_total"
        }
    )
)

# Get the annual totals that we previously calculated
# from the monthly Beneficiaries and Expenditures data.
existing_totals = annual_summary[
    ["fiscal_year", "total_expenditures"]
].copy()

# Compare only FY2018-FY2026 because category_df
# begins in FY2018.
validation_df = category_totals.merge(
    existing_totals,
    on="fiscal_year",
    how="left"
)

# Calculate the difference between the two independently
# extracted annual totals.
#
# A difference of 0 means the two report sections agree exactly.
validation_df["difference"] = (
    validation_df["category_report_total"]
    - validation_df["total_expenditures"]
)

validation_df
```

```python
# ---------------------------------------------------------
# STANDARDIZE CATEGORY NAMES
# ---------------------------------------------------------

# Create a copy so we preserve the original extracted data.
category_clean_df = category_df.copy()

# Standardize only the category labels that we identified
# as inconsistent across fiscal years.
category_name_map = {
    "Mental Health Facilities-DSH Adjustment":
        "Mental Health Facilities-DSH Adjustments",

    "HCBS Waivers (Fee for Service Only)":
        "HCBS Waivers"
}

# Replace the inconsistent labels with standardized names.
category_clean_df["category"] = (
    category_clean_df["category"]
    .replace(category_name_map)
)

# Check that every fiscal year now contains
# the same standardized set of categories.
category_check = (
    category_clean_df
    .groupby("fiscal_year")["category"]
    .nunique()
)

print(category_check)
```

```python
# ---------------------------------------------------------
# COMPARE CATEGORY EXPENDITURES: FY2023 VS FY2026
# ---------------------------------------------------------

# Keep the categories that represent actual expenditure categories.
#
# We exclude:
# - Total Expenditures because it is the overall total, not a category.
# - KDHE and KDADS KanCare because they are components of Managed Care.
#   Including them alongside Managed Care would double-count spending.
analysis_categories = category_clean_df[
    ~category_clean_df["category"].isin([
        "Total Expenditures",
        "KDHE (KanCare)",
        "KDADs (KanCare)"
    ])
].copy()

# Select FY2023 as our starting point.
fy2023 = (
    analysis_categories[
        analysis_categories["fiscal_year"] == 2023
    ][["category", "annual_expenditures"]]
    .rename(
        columns={
            "annual_expenditures": "fy2023_expenditures"
        }
    )
)

# Select FY2026 as our ending point.
fy2026 = (
    analysis_categories[
        analysis_categories["fiscal_year"] == 2026
    ][["category", "annual_expenditures"]]
    .rename(
        columns={
            "annual_expenditures": "fy2026_expenditures"
        }
    )
)

# Match the same categories between the two fiscal years.
category_change = fy2023.merge(
    fy2026,
    on="category",
    how="inner"
)

# Calculate the absolute dollar change.
category_change["dollar_change"] = (
    category_change["fy2026_expenditures"]
    - category_change["fy2023_expenditures"]
)

# Calculate percentage change.
#
# We use .where() to avoid dividing by zero if a category
# happened to have no expenditures in FY2023.
category_change["percent_change"] = (
    (
        category_change["dollar_change"]
        / category_change["fy2023_expenditures"]
    ) * 100
).where(
    category_change["fy2023_expenditures"] != 0
)

# Sort from the largest increase to the largest decrease.
category_change = category_change.sort_values(
    "dollar_change",
    ascending=False
).reset_index(drop=True)

category_change
```
