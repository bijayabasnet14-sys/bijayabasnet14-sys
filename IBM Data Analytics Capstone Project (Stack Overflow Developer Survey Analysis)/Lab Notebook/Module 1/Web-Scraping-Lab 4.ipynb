## Extract information from the given web site
You will extract the data from the below web site: <br> 


```python
#this url contains the data you need to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

The data you need to scrape is the **name of the programming language** and **average annual salary**.<br> It is a good idea to open the url in your web broswer and study the contents of the web page before you start to scrape.


Import the required libraries


```python
# Your code here
from bs4 import BeautifulSoup # this module helps in web scrapping.
import requests  # this module helps us to download a webpage
```

Download the webpage at the url


```python
#your code goes here
data  = requests.get(url).text 
```

Create a soup object


```python
#your code goes here
soup = BeautifulSoup(data,"html.parser")  # create a soup object using the variable 'data'
```

Scrape the `Language name` and `annual average salary`.


```python
#your code goes here
# your code goes here
import pandas as pd

# Find the html table on the webpage
table = soup.find('table')

# Create an empty list to store row elements
popular_languages = []

# Loop through all rows in the table
for row in table.find_all('tr'):
    # Get all column fields in the current row
    cols = row.find_all('td')
    
    # Column index 1 contains Language, Index 3 contains Salary
    language_name = cols[1].getText()
    average_salary = cols[3].getText()
    
    # Append the extracted row list to our main container
    popular_languages.append([language_name, average_salary])

# Convert the populated list cleanly into a Pandas DataFrame
# Use the first row values as headers, then remove it from the data rows
df = pd.DataFrame(popular_languages[1:], columns=popular_languages[0])
print(df)
```

Save the scrapped data into a file named *popular-languages.csv*


```python
# your code goes here
df.to_csv("popular-languages.csv", index=False)
print("File 'popular-languages.csv' has been created successfully!")
```
