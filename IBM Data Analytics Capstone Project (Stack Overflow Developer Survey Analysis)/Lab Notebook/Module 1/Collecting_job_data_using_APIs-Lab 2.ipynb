```python
api_url="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"
response = requests.get(api_url) # Call the API using the get method and store the
                                # output of the API call in a variable called response.
if response.ok:             # if all is well() no errors, no network timeouts)
    data = response.json()  # store the result in json format in a variable called data
                            # the variable data is of type dictionary.
```

```python
print(data)   # print the data just to check the output or for debugging
```

```python
import json

def get_number_of_jobs_T(technology):
    number_of_jobs = 0
    
    # Open the local jobs.json file shown in your file explorer
    with open("jobs.json", "r", encoding="utf-8") as file:
        data = json.load(file)
        
        for job in data:
            # Safely fetch the 'Key Skills' string
            skills = job.get("Key Skills", "")
            
            # Split the pipe-separated skills and clean up spaces
            skill_list = [s.strip().lower() for s in skills.split("|")]
            
            # Check if the requested technology matches any skill
            if technology.lower() in skill_list:
                number_of_jobs += 1
                
    return technology, number_of_jobs
```

Calling the function for Python and checking if it works.


```python
get_number_of_jobs_T("Python")
```

#### Write a function to find number of jobs in US for a location of your choice


```python
import json

def get_number_of_jobs_L(location):
    number_of_jobs = 0
    
    # Open the local jobs.json file in your directory
    with open("jobs.json", "r", encoding="utf-8") as file:
        data = json.load(file)
        
        for job in data:
            # Safely fetch the 'Location' string from the data
            job_location = job.get("Location", "")
            
            # Clean spaces and do a precise case-insensitive match
            if job_location.strip().lower() == location.lower():
                number_of_jobs += 1
                
    return location, number_of_jobs
```

Call the function for Los Angeles and check if it is working.


```python
# Call the function for Los Angeles and check if it is working
location, number_of_jobs = get_number_of_jobs_L('Los Angeles')

# Print the results to verify it works
print(f"Location: {location}")
print(f"Number of Jobs: {number_of_jobs}")
```

#### In the similar way, you can try for below given key skills and results  can be stored in an excel sheet.


Collect the number of job postings for the following key skills using the API:

*   C
*   C#
*   C++
*   Java
*   JavaScript
*   Python
*   Scala
*   Oracle
*   SQL Server
*   MySQL Server
*   PostgreSQL
*   MongoDB


```python
# your code goes here
# Import the required operational libraries
from openpyxl import Workbook
import requests

wb_tech = Workbook()
ws_tech = wb_tech.active
ws_tech.title = "Technology Job Postings"

# Append the required data columns headers
ws_tech.append(["Technology", "Number of Jobs"])

# Target programming skills payload array as requested
technologies = [
    "C", "C#", "C++", "Java", "JavaScript", "Python", 
    "Scala", "Oracle", "SQL Server", "MySQL Server", "PostgreSQL", "MongoDB"
]

# Query the API server dynamically for each key skill requirement
for tech in technologies:
    # Filter the request payload via the specific "Key Skills" field string
    payload = {"Key Skills": tech}
    
    # Query using your pre-configured lab server string variable
    response = requests.get(api_url, params=payload)
    
    if response.ok:
        data = response.json()
        number_of_jobs = len(data)
    else:
        number_of_jobs = 0
        
    # Write the calculated row tracking information down the active spreadsheet tab
    ws_tech.append([tech, number_of_jobs])

# Export structural metrics to the requested target file storage name
wb_tech.save("job-postings-technology.xlsx")

print("Spreadsheet 'job-postings-technology.xlsx' created and saved successfully!")
```
