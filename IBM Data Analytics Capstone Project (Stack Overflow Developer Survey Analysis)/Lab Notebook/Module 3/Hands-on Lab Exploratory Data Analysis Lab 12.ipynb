#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
## Write your code here
# Identify critical columns to process
target_cols = ['Employment', 'JobSat', 'RemoteWork']

# Fill missing entries with the most frequent value (mode) of each respective column
for col in target_cols:
    col_mode = df[col].mode()[0]
    df[col] = df[col].fillna(col_mode)

# Verify that there are no remaining missing values left in these columns
print("Missing values remaining after imputation:")
print(df[target_cols].isnull().sum())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Clean and convert YearsCodePro to numeric variables safely
df['YearsCodePro_num'] = pd.to_numeric(
    df['YearsCodePro'].replace({'Less than 1 year': 0, 'More than 50 years': 50}), 
    errors='coerce'
).fillna(0)

# 2. Define experience range category grouping function
def assign_experience_range(years):
    if years <= 5:
        return '0–5 years'
    elif years <= 10:
        return '5–10 years'
    elif years <= 20:
        return '10–20 years'
    else:
        return '>20 years'

df['Experience_Range'] = df['YearsCodePro_num'].apply(assign_experience_range)

# 3. Explicitly convert the raw numeric satisfaction scores to a float data type
df['JobSat_numeric'] = pd.to_numeric(df['JobSat'], errors='coerce')

# 4. Group by ranges and calculate the median satisfaction score
satisfaction_by_exp = df.groupby('Experience_Range')['JobSat_numeric'].median().reset_index()

# Sort groups chronologically
order_map = {'0–5 years': 0, '5–10 years': 1, '10–20 years': 2, '>20 years': 3}
satisfaction_by_exp['sort_idx'] = satisfaction_by_exp['Experience_Range'].map(order_map)
satisfaction_by_exp = satisfaction_by_exp.sort_values('sort_idx').dropna(subset=['JobSat_numeric'])

print("Calculated Median Job Satisfaction Scores (1-10 Scale):")
print(satisfaction_by_exp[['Experience_Range', 'JobSat_numeric']])

# 5. Visualize the numerical distribution cleanly
plt.figure(figsize=(10, 5))
bars = plt.barh(
    satisfaction_by_exp['Experience_Range'], 
    satisfaction_by_exp['JobSat_numeric'], 
    color='#2ecc71', 
    edgecolor='black', 
    height=0.6
)

# Add numeric value labels on top of each bar for perfect scannability
for bar in bars:
    width = bar.get_width()
    plt.text(
        width + 0.1, 
        bar.get_y() + bar.get_height()/2, 
        f'{int(width)}', 
        va='center', 
        ha='left', 
        fontweight='bold'
    )

plt.title('Median Job Satisfaction Score by Experience Level', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Median Satisfaction Rating Score (1 to 10 Scale)', fontsize=12)
plt.ylabel('Professional Coding Experience Ranges', fontsize=12)
plt.xlim(0, 11)  # Expand limits slightly past 10 to give the labels room

plt.tight_layout()
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
## Write your code here
import matplotlib.pyplot as plt

# 1. Count the occurrences of each unique response category in JobSat
jobsat_counts = df['JobSat'].value_counts()

# 2. Generate a clean horizontal bar layout
plt.figure(figsize=(10, 5))
plt.barh(jobsat_counts.index, jobsat_counts.values, color='#4a90e2', edgecolor='black')

# 3. Add descriptive title metadata and directional anchors
plt.title('Overall Distribution of Job Satisfaction (JobSat)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Respondents', fontsize=12)
plt.ylabel('Satisfaction Rating Level', fontsize=12)
plt.gca().invert_yaxis()  # Places highest frequency tracking categories at the top

plt.tight_layout()
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
import matplotlib.pyplot as plt

# Print column names to check if a specific "DevType" or "JobRole" column exists
print("Available columns to isolate job roles:")
job_col = next((c for c in df.columns if c.lower() in ['devtype', 'jobrole', 'employment']), 'Employment')
print(f"Using column '{job_col}' for job role classification analysis.\n")

# Cross-tabulate remote work preferences by employment type and job roles
cross_tab = pd.crosstab(
    index=df[job_col], 
    columns=df['RemoteWork'], 
    margins=True, 
    margins_name="Total"
)
print("Cross-Tabulation Matrix (Job Role/Employment vs Remote Work Preference):")
print(cross_tab)

#\Create a percentage-based cross-tabulation to isolate true preference ratios
cross_tab_pct = pd.crosstab(
    index=df[job_col], 
    columns=df['RemoteWork'], 
    normalize='index'
) * 100

#Visualize the remote work distribution using a clustered horizontal count plot
plt.figure(figsize=(12, 6))
cross_tab_pct.plot(kind='barh', stacked=False, ax=plt.gca(), cmap='viridis', edgecolor='black')

# 5. Add titles and layout structural anchors
plt.title('Remote Work Preferences Distribution by Role / Employment Track', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Percentage of Respondents (%)', fontsize=12)
plt.ylabel('Job Role Classification / Employment Type', fontsize=12)
plt.legend(title='Remote Work Status', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adjust left margin so long role names fit comfortably without wrapping warnings
plt.subplots_adjust(left=0.3)
plt.tight_layout()
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
import matplotlib.pyplot as plt
import pandas as pd

#  Clean up whitespaces and case-standardize the Country column to avoid blank filters
df['Country_clean'] = df['Country'].astype(str).str.strip().str.title()

# Print diagnostic verification info to console
print("Top 5 country name strings found in your dataset:")
print(df['Country_clean'].value_counts().head(5))
print("-" * 50)

#  Filter for a verified country present in the leaderboard (fallback to first available if needed)
target_country = "United States"
if target_country not in df['Country_clean'].values:
    target_country = df['Country_clean'].value_counts().idxmax()

print(f"Filtering dataset specifically for region: '{target_country}'")
region_df = df[df['Country_clean'] == target_country].dropna(subset=['LanguageHaveWorkedWith'])

#  Handle multiple string delimiters safely (semicolons, commas, or pipes)
# This prevents string extraction parsing drops
region_df['Language_Cleaned'] = region_df['LanguageHaveWorkedWith'].astype(str).str.replace(',', ';').str.replace('|', ';', regex=False)

#  Explode and tally separate languages
lang_series = region_df['Language_Cleaned'].str.split(';').explode()
lang_series = lang_series.str.strip()  # Clean whitespace around individual language items
top_languages = lang_series.value_counts()

# Remove empty or non-standard null string entries from tracking charts
top_languages = top_languages[top_languages.index.str.lower() != 'nan'].head(10)

#  Visualize if records populate
if not top_languages.empty:
    plt.figure(figsize=(10, 5))
    plt.barh(top_languages.index, top_languages.values, color='#8e44ad', edgecolor='black')
    
    plt.title(f'Top 10 Most Popular Programming Languages in {target_country}', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('Number of Responses (Users)', fontsize=12)
    plt.ylabel('Programming Language', fontsize=12)
    plt.gca().invert_yaxis()
    
    plt.tight_layout()
    plt.show()
else:
    print("\n[Warning] Language series is resolving as empty. Check raw values:")
    print(df['LanguageHaveWorkedWith'].head(10))

```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
## Write your code here
import matplotlib.pyplot as plt
import numpy as np

# 1. Strip and clean YearsCodePro to continuous numbers safely
df['YearsCodePro_num'] = pd.to_numeric(
    df['YearsCodePro'].replace({'Less than 1 year': 0, 'More than 50 years': 50}), 
    errors='coerce'
).fillna(0)

# 2. Clean your satisfaction points column to numbers
df['JobSatPoints_num'] = pd.to_numeric(df['JobSatPoints_1'], errors='coerce')

# 3. Calculate a basic correlation coefficient matrix
correlation = df['YearsCodePro_num'].corr(df['JobSatPoints_num'])
print(f"The calculated Pearson correlation coefficient is: {round(correlation, 4)}")

# 4. Use small statistical random noise (jittering) to prevent point overlapping on the chart
jitter_x = df['YearsCodePro_num'] + np.random.uniform(-0.3, 0.3, size=len(df))
jitter_y = df['JobSatPoints_num'] + np.random.uniform(-0.3, 0.3, size=len(df))

plt.figure(figsize=(10, 6))
plt.scatter(jitter_x, jitter_y, alpha=0.3, color='#e67e22', s=15, edgecolor='none')
plt.title('Scatter Plot: Coding Experience vs. Job Satisfaction', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Years of Professional Coding Experience (YearsCodePro)', fontsize=12)
plt.ylabel('Job Satisfaction Score (JobSatPoints_1)', fontsize=12)

plt.tight_layout()
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here
import matplotlib.pyplot as plt

# 1. Generate a standardized cross-tabulation normalization matrix 
edu_employment_pct = pd.crosstab(
    index=df['EdLevel'], 
    columns=df['Employment'], 
    normalize='index'
) * 100

print("Cross-Tabulation Percentage Profile Summary Matrix (%):")
print(edu_employment_pct.round(2))

# 2. Visualize using an unstacked horizontal group chart
plt.figure(figsize=(12, 6))
edu_employment_pct.plot(kind='barh', stacked=True, ax=plt.gca(), cmap='plasma', edgecolor='black')

plt.title('Relationship Map: Educational Background vs Employment Configurations', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Composition Percentage Breakdown within Degree Tier (%)', fontsize=12)
plt.ylabel('Highest Academic Educational Level (EdLevel)', fontsize=12)
plt.legend(title='Employment Class Status', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.subplots_adjust(left=0.35)
plt.tight_layout()
plt.show()
```
