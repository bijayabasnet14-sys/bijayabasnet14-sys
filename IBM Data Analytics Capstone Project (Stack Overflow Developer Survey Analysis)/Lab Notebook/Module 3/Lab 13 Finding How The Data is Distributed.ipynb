### Step 1: Import Libraries and Load Data


- Import the `pandas`, `matplotlib.pyplot`, and `seaborn` libraries.


- You will begin with loading the dataset. You can use the pyfetch method if working on JupyterLite. Otherwise, you can use pandas' read_csv() function directly on their local machines or cloud environments.


```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Display the first few rows of the dataset
df.head()
```

### Step 2: Examine the Structure of the Data


- Display the column names, data types, and summary information to understand the data structure.

- Objective: Gain insights into the dataset's shape and available variables.


```python
## Write your code here
# Print the shape of the dataset (Rows, Columns)
print(f"Dataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns\n")

#  Display column names, non-null counts, and data types
print("--- Concise Summary Information ---")
df.info()

# Preview the first 5 rows to understand the variable layout
print("\n--- First 5 Rows Preview ---")
df.head()
```

### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
## Write your code here
# 1. Identify missing values across all columns
print("Missing values before cleaning:")
print(df[['Employment', 'JobSat', 'YearsCodePro']].isnull().sum())

# 2. Impute categorical variables with their most frequent value (mode)
df['Employment'] = df['Employment'].fillna(df['Employment'].mode()[0])
df['JobSat'] = df['JobSat'].fillna(df['JobSat'].mode()[0])

# 3. Clean and impute numerical experience data with the median
df['YearsCodePro'] = df['YearsCodePro'].replace({'Less than 1 year': 0, 'More than 50 years': 50})
df['YearsCodePro_num'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df['YearsCodePro_num'] = df['YearsCodePro_num'].fillna(df['YearsCodePro_num'].median())

print("\nMissing values after cleaning:")
print(df[['Employment', 'JobSat', 'YearsCodePro_num']].isnull().sum())
```

### Step 4: Analyze Key Columns


- Examine key columns such as `Employment`, `JobSat` (Job Satisfaction), and `YearsCodePro` (Professional Coding Experience).

- **Instruction**: Calculate the value counts for each column to understand the distribution of responses.



```python
## Write your code here
# Calculate and print value frequency distributions for each core variable
print("=== Employment Type Frequency ===")
print(df['Employment'].value_counts())

print("\n=== Job Satisfaction (JobSat) Frequency ===")
print(df['JobSat'].value_counts())

print("\n=== Professional Coding Experience (YearsCodePro) Summary ===")
print(df['YearsCodePro_num'].describe())
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
## Write your code here
import matplotlib.pyplot as plt

# 1. Tally satisfaction frequencies
jobsat_data = df['JobSat'].value_counts()

# 2. Generate a pie chart visualization
plt.figure(figsize=(8, 8))
plt.pie(
    jobsat_data.values, 
    labels=jobsat_data.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=['#2ecc71', '#3498db', '#f1c40f', '#e67e22', '#e74c3c']
)

plt.title('Distribution Profile of Overall Job Satisfaction', fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()

# 3. Text Interpretation
print("\n--- Visual Trend Interpretation ---")
print("1. Dominant Cohort: The vast majority of survey respondents report positive job satisfaction sentiments.")
print("2. Neutral Threshold: A smaller, distinct group remains completely neutral ('Neither satisfied nor dissatisfied').")
print("3. Actionable Takeaway: The skew toward satisfaction implies favorable industry career retention metrics overall.")
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
## Write your code here
import matplotlib.pyplot as plt
import pandas as pd

# Split, explode, and count individual languages for current languages
worked_with = df['LanguageHaveWorkedWith'].str.split(';').explode().str.strip().value_counts()

# Split, explode, and count individual languages for desired languages
want_to_work = df['LanguageWantToWorkWith'].str.split(';').explode().str.strip().value_counts()

# Combine the counts into a single comparison DataFrame for the top 10 most popular current languages
lang_comparison = pd.DataFrame({'Current': worked_with, 'Desired': want_to_work}).fillna(0)
top_10_langs = lang_comparison.loc[worked_with.head(10).index]

# Visualize using a grouped bar chart
plt.figure(figsize=(12, 6))
top_10_langs.plot(kind='barh', ax=plt.gca(), width=0.8, color=['#3498db', '#e67e22'], edgecolor='black')
plt.title('Programming Languages: Currently Using vs. Want to Learn', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Respondents', fontsize=12)
plt.ylabel('Programming Language', fontsize=12)
plt.gca().invert_yaxis()
plt.legend(title='Language Metric')
plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
## Write your code here
import matplotlib.pyplot as plt
import pandas as pd

# Identify the top 5 most common countries in the dataset to filter the view cleanly
top_countries = df['Country'].value_counts().head(5).index
filtered_df = df[df['Country'].isin(top_countries)]

# Create a percentage cross-tabulation table to compare accurately across different sized regions
remote_by_region = pd.crosstab(filtered_df['Country'], filtered_df['RemoteWork'], normalize='index') * 100

# Visualize using a grouped horizontal bar plot
plt.figure(figsize=(12, 6))
remote_by_region.plot(kind='barh', ax=plt.gca(), cmap='viridis', edgecolor='black', width=0.8)
plt.title('Remote Work Configuration Distribution Across Top Regions', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Percentage of Regional Respondents (%)', fontsize=12)
plt.ylabel('Country / Region', fontsize=12)
plt.legend(title='Remote Work Model', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.subplots_adjust(left=0.25)
plt.tight_layout()
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
import pandas as pd

# 1. Clean and convert YearsCodePro experience strings into floating numbers safely
df['YearsCodePro_num'] = pd.to_numeric(
    df['YearsCodePro'].replace({'Less than 1 year': 0, 'More than 50 years': 50}), 
    errors='coerce'
).fillna(0)

# 2. Check if JobSat contains text or numbers and convert to continuous metrics
df['JobSat_clean'] = df['JobSat'].astype(str).str.strip().str.lower()
jobsat_text_mapping = {
    'very satisfied': 5, 'slightly satisfied': 4, 'neither satisfied nor dissatisfied': 3,
    'slightly dissatisfied': 2, 'very dissatisfied': 1
}

if df['JobSat_clean'].isin(jobsat_text_mapping.keys()).any():
    df['JobSat_numeric'] = df['JobSat_clean'].map(jobsat_text_mapping)
else:
    df['JobSat_numeric'] = pd.to_numeric(df['JobSat'], errors='coerce')

# 3. Use the built-in Pearson method to calculate correlation without needing scipy
pearson_corr = df['YearsCodePro_num'].corr(df['JobSat_numeric'], method='pearson')
print(f"The calculated Pearson correlation coefficient is: {round(pearson_corr, 4)}")

# 4. Interpretation check printout
if abs(pearson_corr) < 0.1:
    print("Interpretation: There is a negligible/very weak relationship between coding tenure and reported job satisfaction.")
else:
    print("Interpretation: There is a notable statistical relationship between experience and satisfaction levels.")
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
## Write your code here
import matplotlib.pyplot as plt
import pandas as pd

# 1. Create a percentage cross-tabulation normalized by row (Education Level)
edu_employment_cross = pd.crosstab(df['EdLevel'], df['Employment'], normalize='index') * 100
print("=== Cross-Tabulation Summary Percentage Matrix (%) ===")
print(edu_employment_cross.round(2))

# 2. Visualize the layout with a stacked horizontal bar plot
plt.figure(figsize=(12, 6))
edu_employment_cross.plot(kind='barh', stacked=True, ax=plt.gca(), cmap='tab10', edgecolor='black')

plt.title('Employment Configurations Across Academic Education Levels (EdLevel)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Composition Percentage Breakdown within Education Tier (%)', fontsize=12)
plt.ylabel('Highest Completed Education Level', fontsize=12)
plt.legend(title='Employment Class Type', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.subplots_adjust(left=0.35)
plt.tight_layout()
plt.show()
```
