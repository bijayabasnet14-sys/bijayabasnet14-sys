### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
## Write your code here
# Count the number of duplicate rows in the dataset
print("Number of duplicate rows:", df.duplicated().sum())
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
## Write your code here
# Drop absolute duplicates and verify the new size
df.drop_duplicates(inplace=True)
print("Dataset size after duplicate removal:", df.shape)
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
## Write your code here
# Show all columns containing missing values along with their count
missing_summary = df.isnull().sum()
print(missing_summary[missing_summary > 0])
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
## Write your code here
# Calculate missing rows specifically for RemoteWork
print("Missing rows in 'RemoteWork' column:", df['RemoteWork'].isnull().sum())
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
## Write your code here
# Display the breakdown frequencies of responses in the RemoteWork column
df['RemoteWork'].value_counts()
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
## Write your code here
# Identify the category with the highest frequency count (the mode)
remotework_mode = df['RemoteWork'].value_counts().idxmax()
print("The most frequent (majority) value in 'RemoteWork' is:", remotework_mode)
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
# Impute the missing empty rows using the identified mode value
df['RemoteWork'] = df['RemoteWork'].fillna(remotework_mode)

# Verify that there are no remaining missing values left in the column
print("Remaining missing entries in 'RemoteWork':", df['RemoteWork'].isnull().sum())
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
## Write your code here
# 1. Look for column keys containing compensation keyword variants
comp_cols = [col for col in df.columns if 'comp' in col.lower()]
print("Identified compensation-related columns:", comp_cols)

# 2. Describe the distribution of these target columns
if comp_cols:
    print("\nStatistical distribution metrics summary:")
    print(df[comp_cols].describe())
else:
    print("\nNo specific explicit compensation column variants found.")
```
