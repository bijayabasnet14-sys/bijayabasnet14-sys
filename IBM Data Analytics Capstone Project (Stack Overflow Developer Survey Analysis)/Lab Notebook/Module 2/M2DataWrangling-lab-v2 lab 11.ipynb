### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Write your code here
# Display a concise summary of the dataframe including dtypes, counts, and non-null values
df.info()

# Alternatively, check exact missing value counts per column
print("\nMissing values per column:")
print(df.isnull().sum())
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Write your code here
# Generate descriptive statistics summarizing central tendency, dispersion, and shape
df.describe()
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
# Inspect unique values in Country to find spelling errors, duplicates, or formatting issues
print("Unique countries:")
print(df['Country'].unique())

# Show value frequencies to see if any outliers or anomalies stand out
print("\nCountry value counts:")
print(df['Country'].value_counts().head(20))
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
# Clean whitespace and standardize capitalization anomalies first
df['Country'] = df['Country'].astype(str).str.strip().str.title()

# Map common variations to a unified format if needed (adjust strings based on your 3.1 findings)
country_mapping = {
    'United States Of America': 'United States',
    'Usa': 'United Kingdom',
    'Uk': 'United Kingdom'
}
df['Country'] = df['Country'].replace(country_mapping)

print("Standardized unique countries snippet:")
print(df['Country'].unique()[:10])
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
# Use pandas get_dummies to create a one-hot encoded representation matrix
df_encoded = pd.get_dummies(df, columns=['Employment'], prefix='Emp', drop_first=False)

# Preview the newly created boolean layout binary columns
print(df_encoded.filter(like='Emp_').head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
# Sort all columns descending based on missing value volumes
highest_missing = df.isnull().sum().sort_values(ascending=False)

print("Top 10 columns with the highest number of missing values:")
print(highest_missing.head(10))
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
# Calculate the median to handle right-skewed salary data safely
median_comp = df['ConvertedCompYearly'].median()

# Fill the missing values with the calculated median
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(median_comp)
print("Missing values remaining in ConvertedCompYearly:", df['ConvertedCompYearly'].isnull().sum())
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
# Identify the most frequent category value (mode)
remotework_mode = df['RemoteWork'].mode()[0]

# Fill the missing gaps with the mode category
df['RemoteWork'] = df['RemoteWork'].fillna(remotework_mode)
print("Missing values remaining in RemoteWork:", df['RemoteWork'].isnull().sum())
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
# Extract the min and max benchmarks
min_val = df['ConvertedCompYearly'].min()
max_val = df['ConvertedCompYearly'].max()

# Apply the Min-Max linear transformation formula mapping data into a 0 to 1 scaling range
df['ConvertedCompYearly_MinMax'] = (df['ConvertedCompYearly'] - min_val) / (max_val - min_val)
print("Min-Max range mapping bounds verified:", df['ConvertedCompYearly_MinMax'].min(), "to", df['ConvertedCompYearly_MinMax'].max())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np

# Apply numpy log1p (log of 1 + x) to compress right-skewed salary fields and handle zeroes safely
df['ConvertedCompYearly_Log'] = np.log1p(df['ConvertedCompYearly'])
print("Log transformation complete. Preview:")
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_Log']].head())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
# Ensure YearsCodePro is numeric, replacing any text entries like 'Less than 1 year' with 0 safely
df['YearsCodePro_num'] = pd.to_numeric(df['YearsCodePro'].replace({'Less than 1 year': 0, 'More than 50 years': 50}), errors='coerce').fillna(0)

# Build a mapping logic function to classify professional seniority cohorts 
def classify_experience(years):
    if years <= 2:
        return 'Junior'
    elif years <= 5:
        return 'Mid-level'
    else:
        return 'Senior'

# Apply the functional transformation row-by-row
df['ExperienceLevel'] = df['YearsCodePro_num'].apply(classify_experience)
print("Experience level categories distribution metrics:")
print(df['ExperienceLevel'].value_counts())
```
