### Step 2: Load the Dataset into a DataFrame



load the dataset using pd.read_csv()

### Step 3: Identifying Duplicate Rows


**Task 1: Identify Duplicate Rows**
  1. Count the number of duplicate rows in the dataset.
  2. Display the first few duplicate rows to understand their structure.


```python
## Write your code here
# 1. Count the number of duplicate rows in the dataset
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows before removal:", duplicate_count)

# 2. Display the first few duplicate rows to understand their structure
df[df.duplicated()].head()
```

### Step 4: Removing Duplicate Rows


**Task 2: Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


```python
## Write your code here
# 1. Remove duplicate rows from the dataset using the drop_duplicates() function
df.drop_duplicates(inplace=True)

# 2. Verify the removal by counting the number of duplicate rows after removal
duplicate_count_after = df.duplicated().sum()
print("Number of duplicate rows after removal:", duplicate_count_after)
print("Remaining total dataset rows:", df.shape[0])
```

### Step 5: Handling Missing Values


**Task 3: Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


```python
## Write your code here
# 1. Identify missing values for all columns in the dataset
missing_data_summary = df.isnull().sum()
print("Missing values per column summary:")
print(missing_data_summary[missing_data_summary > 0]) # Shows only columns with missing data

# 2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value
most_frequent_value = df['EdLevel'].mode()[0]
print(f"\nThe most frequent value (mode) for EdLevel is: '{most_frequent_value}'")

# Impute the missing values with the mode
df['EdLevel'].fillna(most_frequent_value, inplace=True)
print("Missing values remaining in EdLevel column:", df['EdLevel'].isnull().sum())
```

### Step 6: Normalizing Compensation Data


**Task 4: Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


```python
# 1. Print all columns to verify exactly how it's spelled
print("Your dataset columns are:")
print(df.columns.tolist())

# 2. Clean up any accidental whitespaces in column headers
df.columns = df.columns.str.strip()

# 3. Safe normalization loop that handles missing column names gracefully
annual_salaries = []

if 'CompFreq' in df.columns and 'CompTotal' in df.columns:
    for index, row in df.iterrows():
        freq = str(row['CompFreq']).strip().lower()
        total = row['CompTotal']
        
        if pd.isna(freq) or pd.isna(total):
            annual_salaries.append(None)
        elif freq == 'yearly':
            annual_salaries.append(total * 1)
        elif freq == 'monthly':
            annual_salaries.append(total * 12)
        elif freq == 'weekly':
            annual_salaries.append(total * 52)
        else:
            annual_salaries.append(None)
            
    df['NormalizedComp'] = annual_salaries
    print("\nSuccess! 'NormalizedComp' column created.")
    print(df[['CompFreq', 'CompTotal', 'NormalizedComp']].head())
else:
    print("\n[ERROR] 'CompFreq' or 'CompTotal' are still missing. Make sure to scroll to the top of your notebook and re-run cell [4] to load the dataset correctly!")
```
